```python
import csv
age = []
sex = [] 
bmi = []
has_children = []
smoker = []
region = []
charges = []
#with open('insurance.csv') as insurance_csv:
    #ins_reader = csv.DictReader(insurance_csv)
    #for row in ins_reader:
        #age.append(row['age'])
        #sex.append(row['sex'])
        #bmi.append(row['bmi'])
#Function to load data into lists
def import_data(lst, column_name):
    with open('insurance.csv') as insurance_csv:
        ins_reader = csv.DictReader(insurance_csv)
        for row in ins_reader:
           lst.append(row[column_name])
        return lst
import_data(age, 'age')
import_data(sex, 'sex')
import_data(bmi, 'bmi')
import_data(has_children, 'children')
import_data(smoker, 'smoker')
import_data(region, 'region')
import_data(charges, 'charges')
```
