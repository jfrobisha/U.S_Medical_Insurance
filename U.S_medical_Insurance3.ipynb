Importing and Inspecting Dataset : Using Pandas and DataFrames

```python
import pandas as pd
import csv
import numpy as np
import matplotlib.pyplot as plt
ins_df = pd.read_csv('insurance.csv')
print(ins_df.head(10))
print(ins_df.info())
#Change column names for something more suitable
ins_df.columns = ['Age','Sex','BMI','num_of_children','Smoker','Region','Med_Charges($)']
#An easy way for summarising stats . NOTE : If any values are missing then thes values would be incorrect.
ins_df.describe()
#Check for missing values ; In this case none are found.
ins_df.isna().sum()
```

Inspecting charge comparison between Sex and Smoker Status

```python
male_smokers = ins_df[(ins_df.Sex == 'male') & (ins_df.Smoker == 'yes')]
female_smokers = ins_df[(ins_df.Sex == 'female') & (ins_df.Smoker == 'yes')]
male_non_smokers = ins_df[(ins_df.Sex == 'male') & (ins_df.Smoker == 'no')]
female_non_smokers = ins_df[(ins_df.Sex == 'female') & (ins_df.Smoker == 'no')]

def avg_insurance(data):
  return np.average(data['Med_Charges($)'])
diff_in_charges = avg_insurance(male_smokers) - avg_insurance(female_smokers)
print(f'The average charge for Female patients who smoke is ${avg_insurance(female_smokers):,.2f}')
print(f'The average charge for Male patients who smoke is ${avg_insurance(male_smokers):,.2f}')
print(f"On average, female patients who smoke pay ${diff_in_charges:,.2f} less than male smokers.")
print(f'The average charge for Female who do not smoke is ${avg_insurance(female_non_smokers):,.2f}')
print(f'The average charge for Male who do not smoke is ${avg_insurance(male_non_smokers):,.2f}')
print(f"On average, female who do not smoke pay ${avg_insurance(female_non_smokers) - avg_insurance(male_non_smokers) :,.2f} more than male who do not smoke.")
```

Inspecting Charge comparison between smokers and non smokers

```python
smokers = ins_df[ins_df.Smoker == 'yes']
non_smokers = ins_df[ins_df.Smoker == 'no']
print(f'The average charge for people who smoke is ${avg_insurance(smokers):,.2f}')
print(f'The average charge for people who do no smoke is ${avg_insurance(non_smokers):,.2f}')
print(f"On average, people who smoke pay ${avg_insurance(smokers) - avg_insurance(non_smokers):,.2f} more than those who do not.")
```
